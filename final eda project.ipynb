## 6. Data Quality Assessment

**Objective:** To check the raw data for completeness, consistency and accuracy, and then clean it.

**Concept:** **Completeness** = no missing values; **Consistency** = uniform formats and no duplicates; **Accuracy** = values are realistic (e.g. a rating must be between 1 and 5).

**Methodology:** Pandas functions `isnull()`, `duplicated()` and simple filters are applied to `apps_raw` and `reviews_raw`. Cleaning is done on copies.

### 6.1 Completeness
**Objective:** To find how many values are missing in each column.

```python
missing_apps = apps_raw.isnull().sum()
print("Missing values in Apps dataset:")
print(missing_apps[missing_apps > 0])
print("Missing rating (%):", round(missing_apps["Rating"] / len(apps_raw) * 100, 2))
```

```python
missing_reviews = reviews_raw.isnull().sum()
print("Missing values in Reviews dataset:")
print(missing_reviews[missing_reviews > 0])
```

### 6.2 Consistency
**Objective:** To find duplicate records and inconsistent value formats.

```python
print("Duplicate rows in apps:", apps_raw.duplicated().sum())
print("Repeated app names in apps:", apps_raw["App"].duplicated().sum())
print("Duplicate rows in reviews:", reviews_raw.duplicated().sum())
```

```python
print("Reviews values that are not numbers:", pd.to_numeric(apps_raw["Reviews"], errors="coerce").isnull().sum())
print("Size = 'Varies with device':", (apps_raw["Size"] == "Varies with device").sum())
```

### 6.3 Accuracy
**Objective:** To find impossible values (a rating cannot be above 5).

```python
bad_rating = apps_raw[apps_raw["Rating"] > 5]
print("Rows with Rating above 5:", len(bad_rating))
display(bad_rating)
```

```python
print("Observations")
print("- Missing ratings (%):", round(apps_raw["Rating"].isnull().mean() * 100, 1))
print("- Fully duplicated app rows:", apps_raw.duplicated().sum())
print("- Rows with an impossible rating:", len(bad_rating))
```

### 6.4 Cleaning the apps data
**Objective:** To remove the invalid row and the duplicate apps.

```python
apps = apps_raw.drop(index=bad_rating.index)    # remove the corrupted row
apps = apps.drop_duplicates("App")              # keep one row for each app
print("Rows before:", len(apps_raw), "| Rows after:", len(apps))
```

### 6.5 Converting text columns to numbers
**Objective:** To remove symbols such as `$`, `,` and `+` and convert the columns to numbers and dates.

```python
apps["Price"] = apps["Price"].str.replace("$", "", regex=False).astype(float)
apps["Installs"] = apps["Installs"].str.replace(",", "", regex=False).str.replace("+", "", regex=False).astype(int)
apps["Reviews"] = apps["Reviews"].astype(int)
apps["Last Updated"] = pd.to_datetime(apps["Last Updated"])
```

```python
print(apps[["Price", "Installs", "Reviews", "Last Updated"]].dtypes)
```

### 6.6 Cleaning the reviews data
**Objective:** To keep only reviews that have a sentiment and remove repeated reviews.

```python
reviews = reviews_raw.dropna(subset=["Sentiment"])    # keep rows that have a sentiment
reviews = reviews.drop_duplicates()                   # remove repeated reviews
print("Review rows before:", len(reviews_raw), "| Review rows after:", len(reviews))
```

**Interpretation:** Missing ratings, duplicate apps, one impossible rating and text-type numeric columns were found; the cleaned data now has numeric types.

**Decision:** The impossible-rating row was removed (data-entry error); duplicate apps and repeated reviews were removed so nothing is counted twice; missing ratings were kept, not filled; reviews without sentiment were dropped; `$`, `,` and `+` were removed to make numeric columns. `Size` is not used, so it is left unconverted.

**Conclusion:** The data quality issues were identified and handled; the cleaned `apps` and `reviews` DataFrames are used for the rest of the analysis.

## 7. Feature Engineering

**Objective:** To create new columns that describe price type, popularity, rating quality and update year more clearly than the raw columns.

**Concept:** Feature engineering means deriving new variables from existing ones so that patterns are easier to see.

**Methodology:** `np.where()` creates a two-way label; `pd.cut()` converts numbers into labelled bins (each bin includes its right edge); `.dt.year` extracts the year from a date.

### 7.1 Price_Type
**Objective:** To label each app as Free or Paid.

```python
apps["Price_Type"] = np.where(apps["Price"] == 0, "Free", "Paid")
display(apps[["App", "Price", "Price_Type"]].head())
print(apps["Price_Type"].value_counts())
```

### 7.2 Install_Level
**Objective:** To group apps into Low (up to 100,000), Medium (up to 1,000,000) and High (above 1,000,000) installs.

```python
apps["Install_Level"] = pd.cut(apps["Installs"], bins=[-1, 100000, 1000000, np.inf], labels=["Low", "Medium", "High"])
display(apps[["App", "Installs", "Install_Level"]].head())
print(apps["Install_Level"].value_counts())
```

### 7.3 Rating_Level
**Objective:** To group ratings into Poor (up to 2.5), Average (up to 3.5), Good (up to 4.5) and Excellent (up to 5).

```python
apps["Rating_Level"] = pd.cut(apps["Rating"], bins=[0, 2.5, 3.5, 4.5, 5], labels=["Poor", "Average", "Good", "Excellent"])
display(apps[["App", "Rating", "Rating_Level"]].head())
print(apps["Rating_Level"].value_counts())
```

### 7.4 Update_Year
**Objective:** To extract the year in which each app was last updated.

```python
apps["Update_Year"] = apps["Last Updated"].dt.year
display(apps[["App", "Last Updated", "Update_Year"]].head())
print(apps["Update_Year"].value_counts().sort_index())
```

```python
print("Observations")
print("- Paid apps (%):", round((apps["Price_Type"] == "Paid").mean() * 100, 1))
print("- Most common install level:", apps["Install_Level"].value_counts().idxmax())
print("- Most common rating level:", apps["Rating_Level"].value_counts().idxmax())
print("- Year with most updates:", apps["Update_Year"].value_counts().idxmax())
```

**Interpretation:** Each new column turns a raw value into a simple label, and the count tables show how apps are spread across the labels.

**Decision:** Fixed, easy-to-explain thresholds were chosen for the bins. Apps with a missing Rating get no Rating_Level (they stay missing) instead of being guessed.

**Conclusion:** Four new features were created and will support grouping, pivoting and comparison in the next sections.

## 8. Grouping and Aggregation

**Objective:** To summarise apps category by category.

**Concept:** Grouping splits data by a column and calculates a summary for each group. `agg()` returns one row per group, while `transform()` returns one value for every original row.

**Methodology:** `groupby('Category')` with `count`, `mean`, `median`; `transform('mean')` adds the category average back to each app.

### 8.1 Apps per category
**Objective:** To find which categories contain the most apps.

```python
print(apps["Category"].value_counts().head(10))
```

### 8.2 Multiple summaries with agg()
**Objective:** To compute several statistics for every category at once.

```python
category_summary = apps.groupby("Category").agg(
    Apps=("App", "count"), Avg_Rating=("Rating", "mean"), Median_Installs=("Installs", "median"))
display(category_summary.sort_values("Apps", ascending=False).head(10).round(2))
```

### 8.3 Group value for every row with transform()
**Objective:** To compare each app's rating with the average of its own category.

```python
apps["Category_Avg_Rating"] = apps.groupby("Category")["Rating"].transform("mean")
apps["Rating_vs_Category"] = apps["Rating"] - apps["Category_Avg_Rating"]
display(apps[["App", "Category", "Rating", "Category_Avg_Rating", "Rating_vs_Category"]].head())
```

```python
print("Observations")
print("- Category with most apps:", category_summary["Apps"].idxmax())
print("- Category with highest average rating:", category_summary["Avg_Rating"].idxmax())
print("- Category with lowest average rating:", category_summary["Avg_Rating"].idxmin())
```

**Interpretation:** The table shows which categories are largest and how their ratings differ; a positive `Rating_vs_Category` means the app is above its category average.

**Decision:** Median installs are shown beside the mean rating because installs are heavily skewed, and the median is not affected by a few very popular apps.

**Conclusion:** Grouping shows clear differences in size and rating between categories, supporting the objective of comparing categories.

## 9. Pivot Tables and Crosstab

**Objective:** To compare ratings and app counts across categories and price types.

**Concept:** A pivot table summarises a numerical column across two categorical columns; a crosstab counts how many records fall in each combination.

**Methodology:** `pd.pivot_table(values, index, columns, aggfunc)` and `pd.crosstab()` (with `normalize` for percentages and `margins` for totals). The five largest categories are used to keep tables readable.

```python
top5 = apps["Category"].value_counts().head(5).index        # five largest categories
top_apps = apps[apps["Category"].isin(top5)]
rating_pivot = pd.pivot_table(top_apps, values="Rating", index="Category", columns="Price_Type", aggfunc="mean")
display(rating_pivot.round(2))
```

```python
display(pd.crosstab(top_apps["Category"], top_apps["Price_Type"], margins=True))
```

```python
price_share = pd.crosstab(top_apps["Category"], top_apps["Price_Type"], normalize="index") * 100
display(price_share.round(2))
```

```python
print("Observations")
print("- Category with highest share of paid apps:", price_share["Paid"].idxmax())
print("- Its paid share (%):", round(price_share["Paid"].max(), 1))
```

**Interpretation:** The pivot table gives average ratings for free and paid apps in each category, and the crosstab shows how many apps are free or paid.

**Decision:** Percentages (row-wise) were added because raw counts alone cannot be compared across categories of different size.

**Conclusion:** Pivot tables and crosstabs show how pricing and ratings are distributed across the biggest categories.

## 10. Data Reshaping

**Objective:** To demonstrate how the same data can be arranged in wide or long layouts.

**Concept:** Wide format has one column per measure; long format has one row per measure. `melt()` goes wide → long, `pivot()` goes long → wide, `stack()` moves columns into the row index and `unstack()` moves them back.

**Methodology:** A small category-level summary (first 3 categories) is used so the results are easy to read. The original `apps` data is not changed.

```python
wide = apps.groupby("Category")[["Rating", "Installs"]].mean().head(3)
display(wide.round(2))
```

```python
long = wide.reset_index().melt(id_vars="Category", var_name="Metric", value_name="Value")
display(long.round(2))
```

```python
wide_again = long.pivot(index="Category", columns="Metric", values="Value")
display(wide_again.round(2))
```

```python
stacked = wide.stack()
display(stacked.round(2))
display(stacked.unstack().round(2))
```

```python
print("Observations")
print("- Wide table shape:", wide.shape)
print("- Long table shape:", long.shape)
```

**Interpretation:** The long table has 6 rows (one per category-measure pair), and `pivot` and `unstack` return the original table, so no information is lost.

**Decision:** A small summary was used as the demonstration so that reshaping is clear; the main dataset stays in its original layout.

**Conclusion:** Reshaping changes only the layout, not the information, and gives flexibility for different analysis needs.

## 11. Data Integration

**Objective:** To combine app details with user-review information using the common column `App`.

**Concept:** Merging joins tables on a key column. Because one app has many reviews, reviews are first **summarised per app** so that each app stays one row after the join. A **left join** keeps every app, even those without reviews.

**Methodology:** `groupby().agg()` → `merge(how='left', validate='many_to_one', indicator=True)`; also `join()` (index-based), `concat()` (stacking), index alignment and key-conflict handling with `suffixes` and `validate`.

### 11.1 Summarising reviews per app
**Objective:** To convert many reviews per app into one summary row per app.

```python
review_summary = reviews.groupby("App").agg(
    Total_Reviews=("Sentiment", "count"), Avg_Polarity=("Sentiment_Polarity", "mean")).reset_index()
display(review_summary.head())
print("Review rows:", len(reviews), "-> app-level rows:", len(review_summary))
```

### 11.2 merge()
**Objective:** To attach review information to every app using a left join.

```python
merged = apps.merge(review_summary, on="App", how="left", validate="many_to_one", indicator=True)
display(merged[["App", "Category", "Rating", "Total_Reviews", "Avg_Polarity", "_merge"]].head())
print(merged["_merge"].value_counts())
print("Rows before merge:", len(apps), "| after merge:", len(merged))
```

### 11.3 join() and concat()
**Objective:** To show index-based joining and stacking of tables.

```python
joined = apps.set_index("App").join(review_summary.set_index("App"), how="inner")
print("Inner join (only apps that have reviews):", joined.shape)
```

```python
stacked_rows = pd.concat([apps.head(2), apps.tail(2)])    # stack rows one below the other
display(stacked_rows[["App", "Rating"]])
```

### 11.4 Index alignment (controlled demonstration)
**Objective:** To show that pandas matches values by index label, not by position.

```python
s1 = pd.Series([1, 2, 3], index=["a", "b", "c"])
s2 = pd.Series([10, 20, 30], index=["b", "c", "d"])
print(s1 + s2)          # labels a and d have no match, so they become NaN
```

### 11.5 Handling key conflicts
**Objective:** To handle clashing column names and repeated keys safely.

```python
store_rating = apps[["App", "Rating"]]
review_score = review_summary[["App", "Avg_Polarity"]].rename(columns={"Avg_Polarity": "Rating"})
display(store_rating.merge(review_score, on="App", suffixes=("_store", "_review")).head())
```

```python
try:
    apps_raw.merge(review_summary, on="App", validate="one_to_one")
except Exception as error:
    print("Merge stopped because app names are repeated:", type(error).__name__)
```

```python
matched = (merged["_merge"] == "both").sum()
print("Observations")
print("- Apps with review data:", matched, "out of", len(apps))
print("- Apps without review data:", len(apps) - matched)
```

**Interpretation:** Each app now has its review count and average sentiment; the merge indicator shows how many apps matched, and `suffixes` and `validate` handled key conflicts.

**Decision:** Reviews were summarised first so app rows are not duplicated; a left join keeps apps without reviews; `validate='many_to_one'` guards against repeated keys.

**Conclusion:** The two datasets were integrated into `merged`, giving app details and user feedback in one table.

## 14. Empirical Rule

**Objective:** To check how closely app ratings follow the 68–95–99.7 rule.

**Concept:** For an approximately normal distribution, about 68% of values lie within 1 standard deviation of the mean, about 95% within 2, and about 99.7% within 3. In EDA it is used to judge whether data are normal.

**Methodology:** For k = 1, 2, 3: percentage of values with |value − mean| ≤ k × standard deviation. Real ratings are compared with a simulated normal sample (seed 42) and with the theoretical percentages **68%, 95% and 99.7%**.

```python
ratings = apps["Rating"].dropna()
mean, std = ratings.mean(), ratings.std()
for k in [1, 2, 3]:
    print("Real ratings within", k, "SD:", round(((ratings - mean).abs() <= k * std).mean() * 100, 1), "%")
```

```python
np.random.seed(42)
simulated = pd.Series(np.random.normal(50, 10, 10000))
for k in [1, 2, 3]:
    print("Simulated normal within", k, "SD:", round(((simulated - simulated.mean()).abs() <= k * simulated.std()).mean() * 100, 1), "%")
```

```python
print("Observations")
print("- Theory: 68%, 95%, 99.7%")
print("- Skewness of real ratings:", round(ratings.skew(), 2))
print("- Skewness of simulated normal data:", round(simulated.skew(), 2))
```

**Interpretation:** Simulated normal data matches 68–95–99.7 closely, but real ratings are more concentrated and left-skewed, so they are not normal.

**Decision:** The simulation is used only as a labelled controlled demonstration; conclusions about the Play Store come from the real ratings.

**Conclusion:** The empirical rule holds only roughly for ratings; the mismatch itself is useful, because it shows that ratings are not normally distributed.

## 15. Sampling and Central Limit Theorem

**Objective:** To show how sample averages behave when many random samples are drawn.

**Concept:** The **Central Limit Theorem** states that the means of many random samples are approximately normally distributed, even if the original data are not.


**Methodology:** The non-missing ratings are treated as the population. 1,000 random samples of size 30 and 1,000 of size 100 are drawn with pandas `sample()`, and the mean of each sample is stored. **Random seed: 42**, so results are reproducible.

```python
np.random.seed(42)                         # fixed seed -> reproducible samples
population = apps["Rating"].dropna()
print("Population mean:", round(population.mean(), 3))
```

```python
means_30 = [population.sample(30).mean() for i in range(1000)]      # 1000 samples of size 30
means_100 = [population.sample(100).mean() for i in range(1000)]    # 1000 samples of size 100
```

```python
print("Size 30 : mean of sample means =", round(np.mean(means_30), 3), "| std of sample means =", round(np.std(means_30), 3))
print("Size 100: mean of sample means =", round(np.mean(means_100), 3), "| std of sample means =", round(np.std(means_100), 3))
```

```python
print("Theoretical standard error (std / sqrt(n)):")
print("Size 30 :", round(population.std() / np.sqrt(30), 3))
print("Size 100:", round(population.std() / np.sqrt(100), 3))
```

```python
print("Observations")
print("- Skewness of the ratings:", round(population.skew(), 2))
print("- Skewness of sample means (n=30):", round(pd.Series(means_30).skew(), 2))
print("- Skewness of sample means (n=100):", round(pd.Series(means_100).skew(), 2))
```

**Interpretation:** Sample means centre on the true mean and vary less for n=100 than for n=30, close to the theoretical standard error.

**Decision:** A fixed seed was chosen so the experiment can be repeated exactly.

**Conclusion:** The experiment demonstrates the central limit theorem: larger samples give more precise estimates of the average Play Store rating.

## 16. Correlation Analysis

**Objective:** To measure the relationship between Rating, Reviews, Installs and Price.

**Concept:** **Covariance** shows whether two variables move together (its size depends on units). **Correlation** is standardised from −1 to +1. **Pearson** measures linear relationships; **Spearman** uses ranks and is better for skewed data. **Correlation does not prove causation.**

**Methodology:** `DataFrame.cov()` and `DataFrame.corr(method='pearson' / 'spearman')` on the four numeric columns; missing values are ignored pairwise.

```python
corr_data = apps[["Rating", "Reviews", "Installs", "Price"]]
display(corr_data.cov().round(2))
```

```python
pearson = corr_data.corr(method="pearson")
display(pearson.round(3))
```

```python
spearman = corr_data.corr(method="spearman")
display(spearman.round(3))
```

```python
print("Observations")
print("- Rating vs Installs (Pearson):", round(pearson.loc["Rating", "Installs"], 3))
print("- Reviews vs Installs (Pearson):", round(pearson.loc["Reviews", "Installs"], 3))
print("- Reviews vs Installs (Spearman):", round(spearman.loc["Reviews", "Installs"], 3))
```
